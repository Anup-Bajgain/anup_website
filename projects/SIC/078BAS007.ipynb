```python
#Q07-01
adult = input("Are you an adult? , 0 if you are minor: ")
married = input("Are you married? 0 if you are single: ")
if adult == "1" and married == "1":
    print("You are an adult who is married")
elif adult == "1" and married == "0":
    print("You are an adult who is single")
elif adult == "0" and married == "1":
    print("You are a minor who is married")
else:
    print("You are a minor who is single")
```

```python
#Q08-01
for i in range(2, 13):
    if i == 2:
        print("2 : is a prime number")
    count = 0 
    for j in range(2, i):
        if i % j == 0:
            count += 1
    if count == 0:
        print(f"{i} : is a prime number")
    else:
        print(f"{i} : is composite number")
```

```python
#Q09-01
for i in range(100,1000):
    sum = 0
    number = i
    while number > 0:
        digit = number % 10
        sum += digit ** 3
        number //= 10
    if i == sum:
        print(i)
```
