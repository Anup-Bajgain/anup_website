```python
#Q17-1
def distance(x1,y1,x2,y2):
    return ((x1-x2)**2+(y1-y2)**2)**0.5
x1,y1,x2,y2 = input("Enter the co-ordinate: ").split()
print("The distance between two points: ",distance(int(x1),int(y1),int(x2),int(y2)))
```

```python
#Q18-1
def is_palindrome(str):
    if len(str)<=1:
        return True
    if str[0]!=str[-1]:
        return False

    return is_palindrome(str[1:-1])

print(is_palindrome("level"))
print(is_palindrome("notApalindrome"))
```

```python
#Q21-1
class vector2D:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __mul__(self,other):
        return vector2D(self.x*other.x,self.y*other.y)
    def __truediv__(self,other):
        if self.x==0 or self.y==0:
            print("Error")
        return vector2D(self.x/other.x,self.y/other.y)
    def __str__(self):
        return "{},{}".format(self.x,self.y)


v1 = vector2D(10,20)
v2 = vector2D(5,20)
v1xv2=v1*v2
v1divv2 = v1/v2
print("multi:",v1xv2)
print("div:",v1divv2)
        
```
