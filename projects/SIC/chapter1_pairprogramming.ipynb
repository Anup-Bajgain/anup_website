```python
#Q1
print("1) Addition 2) Subtraction 3) Multiplication 4) Division")
n = int(input("Enter the number of the operation : "))
if n<1 or n>4:
    print("Entered an incorrect number")
else:
    print("Enter two numbers for operation.")
    a = int(input())
    b = int(input())
    if n==1:
        print(a,"+",b,"=",a+b)
    elif n==2:
        print(a,"-",b,"=",a-b)
    elif n==3:
        print(a,"*",b,"=",a*b)
    else:
        print(a,"/",b,"=",a/b)
```

```python
#Q2
x,y = map(int,input("Enter x,y coordinates: ").split())
if x>0 and y>0:
    print("1st quadrant")
elif x<0 and y>0:
    print("2nd quadrant")
elif x<0 and y<0:
    print("3rd quadrant")
else:
    print("4th quadrant")
```

```python
#Q3 
print("Welcome to yummy restaurant. Here is the menu.")
print("- Burger(enter b)\n- Chicken(enter c)\n- Pizza(enter p)")
n = input("Choose a menu (enter b,c,p): ")
while n not in 'bcp':
    print("Enter the menu again.")
    n = input("Choose a menu (enter b,c,p): ")
if n=='b':
    print("You chose burger.")
elif n=='c':
    print("You chose chicken.")
else:
    print("You chose pizza.")
```

```python
#Q1
n = int(input("Enter n: "))
for i in range(n):
    if i%2==0:
        for j in range(1,n+1,1):
            print(5*i+j,end=' ')
    else:
        for j in range(n,0,-1):
            print(5*i+j,end=' ')
    print("\n")
```

```python
#Q1
n = int(input("Enter an integer: "))
x = n
rev = 0
while n>0:
    rev = rev*10 + n%10
    n = n//10
if x==rev:
    print(f"{x} is a palindrome number")
else:
    print(f"{x} is not a palindrome number")

```

```python
#Q2
import random
x = random.randint(1,100)
trial = 0
print("Guess a number between 1 to 100")
while True:
    n = int(input("Enter a number:"))
    if n>x:
        print("Lower!")
    elif n<x:
        print("Higher!")
    else:
        print("Congraturlations. Total try = {}".format(trial))
        break
    trial+=1
    
```
