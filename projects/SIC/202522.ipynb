```python
def find_minimum(row,col,triangle):
  if row==len(triangle):
    return 0
  else:
    minimum=min(find_minimum(row+1, col, triangle),
                find_minimum(row+1, col+1, triangle))
    return triangle[row][col]+minimum
```

```python
    
triangle=[
    [2],
    [3, 4],
    [6, 5, 7],
    [4, 1, 8, 3]
]
minimum=find_minimum(0,0,triangle)
print("The minimum sum is:",minimum)
```

```python
triangle_large = [
    [5],
    [9, 6],
    [4, 6, 8],
    [0, 7, 1, 5],
    [3, 8, 2, 4, 6],
    [6, 1, 9, 2, 7, 3]
]

minimum = find_minimum(0, 0, triangle_large)
print("The minimum sum for large triangle is:", minimum)
```

```python
triangle_negative = [
    [2],
    [-3, 4],
    [6, -5, 7],
    [-4, 1, 8, -3]
]
minimum = find_minimum(0, 0, triangle_negative)
print("The minimum sum for negative triangle is:", minimum)

print("The minimum sum for small triangle is:", minimum)
```
