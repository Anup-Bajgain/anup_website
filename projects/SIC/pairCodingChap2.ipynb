```python
#chapter2 unit12 Q1
from collections import Counter
givenTuple = (1,2,5,4,3,2,1,4,7,8,9,9,3,7,3,9)
count = Counter(givenTuple)
maxim = max(count.values())
max_element = [key for key,value in count.items() if value ==maxim]
print("The most frequent element is:",max(max_element))
    

```

```python
#chapter2 unit13 Q1
size = input("Enter the sizes of the array: ").split()
inputs=list(map(int,size))
for n in inputs:
    pattern = [[((i+j)%2) for j in range(n)] for i in range(n)]
    for row in pattern:
        print(" ".join(map(str,row)))
    print("\n")


```

```python
#chapter2 unit14 Q1
items = {"Coffee":7,"Pen":3,"Paper Cup":2,"Milk":1,"Coke":4,"Book":5}
while True:
    menu = int(input("Select menu 1)check stock 2)warehousing 3)release 4)exit: "))
    if menu==1:
        item = input("[check stock] Enter item:")
        if item.capitalize() in items:
            print('Stock:',items[item.capitalize()])
        else:
            print('not in stock')
    elif menu==2:
        item, n = input("[warehousing] Enter item and quantity:").split()
        if item.capitalize() in items:
            items[item.capitalize()]+=int(n)
        else:
            items[item.capitalize()]=int(n)          
    elif menu==3:
        item, n = input("[Release] Enter item and quantity:").split()
        if item.capitalize() in items and int(n)<=items[item.capitalize()]:
            items[item.capitalize()]-=int(n)
        else:
            print("could not release")
    elif menu==4:
        break
```

```python
#chapter2 unit15 Q1
student_tup = (('211101' ,'David Doe' , '010-1234-4500') , ('211102' ,'John Smith','010-2230-6540') , ('211103' ,'Jane Carter', '010-3232-7788'))
studentDict = {}
for i in student_tup:
    student ={}
    student[i[0]] = list(i[1:])
    for key in student:
        print(key," : ",student[key][0])
    studentDict.update(student)
ident = input("Enter student ID number: ")
print(ident, f" student is {studentDict[ident][0]} and phone number is {studentDict[ident][1]}")
```

```python
#chapter2 unit16 Q1
mylist = [(1,2),(4,5),(4,2),(3,1),(9,4)]
value = input("Enter value for a and b: ").split()
a,b =map(int,value)
if (a, b) in mylist:
    print(f"There is  ({a}, {b}) at {mylist.index((a, b)) + 1}th")
elif (b, a) in mylist:
    print(f"There is no ({a}, {b}) but  there is at ({b}, {a}) at {mylist.index((b, a)) + 1}th")
else:
    print(f"There is no {(a,b)} nor {(b,a)}")
```
