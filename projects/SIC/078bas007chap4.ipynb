```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        self.stack.append(item)

    def pop(self):
        if not self.is_empty():
            return self.stack.pop()
        else:
            return "Stack is empty"

    def peek(self):
        if not self.is_empty():
            return self.stack[-1]
        else:
            return "Stack is empty"

    def is_empty(self):
        return len(self.stack) == 0
```

```python
class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, item):
        self.queue.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.queue.pop(0)
        else:
            return "Queue is empty"

    def peek(self):
        if not self.is_empty():
            return self.queue[0]
        else:
            return "Queue is empty"

    def is_empty(self):
        return len(self.queue) == 0
```

```python
#Q22-01
stack = Stack()
stack.push("Banana")
stack.push("Apple")
stack.push("Tomato")
stack.push("Strawberry")
stack.push("Grapes")

stack.pop()
stack.pop()
stack.pop()
stack.pop()
```

```python
#Q24-01
def find_two(nums):
    if not nums:
        return None, None
    max_val = max(nums)
    min_val = min(nums)
    return max_val, min_val

nums = [11, 37, 45, 26, 59, 28, 17, 53]
max_val, min_val = find_two(nums)
print("Maximum value:", max_val)
print("Minimum value:", min_val)
```

```python
#Q24-02
def find_two(nums):
    if not nums:
        return None, None, 0
    
    comparisons = 0
    max_val = nums[0]
    min_val = nums[0]
    
    for num in nums[1:]:
        comparisons += 1
        if num > max_val:
            max_val = num
        comparisons += 1
        if num < min_val:
            min_val = num
    
    return comparisons

nums = [11, 37, 45, 26, 59, 28, 17, 53]
comparisons = find_two(nums)
print("Number of comparisons:", comparisons)
```
