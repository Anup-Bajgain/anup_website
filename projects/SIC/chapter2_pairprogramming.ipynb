```python
#Q1
nums = (1,2,5,4,3,2,1,4,7,8,9,9,3,7,3,9)
count = {}
for num in nums:
    if num in count:
        count[num]+=1
    else:
        count[num]=1
freq = nums[0]

for key,value in zip(count.keys(),count.values()):
    if value > count[freq]:
        freq = key
    if value == count[freq]:
        if key > freq:
             freq = key
print("Given tuples:",nums)
print("The most frequent element:",freq)
```

```python
#Q1
n = int(input("Enter an integer: "))
x = n
rev = 0
while n>0:
    rev = rev*10 + n%10
    n = n//10
if x==rev:
    print(f"{x} is a palindrome number")
else:
    print(f"{x} is not a palindrome number")

```
