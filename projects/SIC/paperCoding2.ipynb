```python
#chapter 2 unit 15 Q1
student_tup = (('211101' ,'David Doe' , '010-1234-4500') , ('211102' ,'John Smith','010-2230-6540') , ('211103' ,'Jane Carter', '010-3232-7788'))
studentDict = {}
for i in student_tup:
    student ={}
    student[i[0]] = list(i[1:])
    print(student)
    studentDict.update(student)

```

```python
#chapter 2 unit 15 Q2
Id = input("Enter the student ID: ")
if Id in studentDict:
    print(f"Name: {studentDict[Id][0]}")
    print(f"Phone Number: {studentDict[Id][1]}")
```
