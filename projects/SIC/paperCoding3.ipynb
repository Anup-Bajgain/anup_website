```python
#chapter 3 unit 17 Q2
def max2(m,n):
    if m>n:
        return m
    else:
        return n
def min2(m,n):
    if m<n:
        return m
    else:
        return n
print("The greater of 100 and 200 is:",max2(100,200))
print("The smaller of 100 and 200 is:",min2(100,200))
```

```python
#chapter 3 unit 17 Q3
def mile2km(mile):
    return mile*1.61
for i in range(1,6):
    print(i, "mile =",mile2km(i),"kilometers")
```

```python
#chapter 3 unit 17 Q4
def cel2fah(cel):
    return cel*9/5+32
for i in range(10,60,10):
    print(i, "degrees Celsius =",cel2fah(i),"degrees Fahrenheit")
```

```python
#chapter 3 unit 18 Q1
def addNumber(n):
    if n==1:
        return 1
    else:
        return n+addNumber(n-1)
print(addNumber(int(input("Enter a number: "))))


    
```

```python
#chapter 3 unit 18 Q2
def power(x,n):
    if n==1:
        return x
    else:
        return x*power(x,n-1)
print(power(int(input("Enter x: ")),int(input("Enter n: "))))
```
