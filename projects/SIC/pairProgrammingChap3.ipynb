## Unit 17

```python
def mean3(lst):
    return (lst[0]+lst[1]+lst[2])/3
def min3(lst):
    small=lst[0]
    for i in lst:
        if i < small:
            small =i
    return small

def max3(lst):
    large=lst[0]
    for i in lst:
        if i > large:
            large =i
    return large

integer = input("Enter three number: ").split()
lst = list(map(int,integer))
print(f"The average value of {lst} is {mean3(lst)}")
print(f"The maximum value of {lst} is {max3(lst)}")
print(f"The minimum value of {lst} is {min3(lst)}")
```

## Unit 18

```python
def euler(n):
    if n <=1:
        return 2
    else:
        return (1/factorial(n)) + (euler(n-1)) 
def factorial(n):
    if n==1:
        return 1
    else:
        return n*factorial(n-1)
print("{:.5f}".format(euler(20)))
    
```

```python
scores = [100, 90, 95, 90, 80, 70, 0, 80, 90, 90, 0, 90, 100, 75, 20, 30, 50, 90]
print("The number of total student is ", int(len(scores)/3))
lst = []
for i in range(int(len(scores)/3)):
    idvl = []
    for j in range(3):
        idvl.append(scores[i*3+j])
    if 0 in idvl:
        continue
    else:
        lst.append(idvl)
print(f"The number of student with valid score is {len(lst)}.\n {lst}")
```

```python
class Student:
    def __init__(self, name, student_id, eng_quiz, math_quiz, science_quiz):
        self.name = name
        self.student_id = student_id
        self.eng_quiz = eng_quiz
        self.math_quiz = math_quiz
        self.science_quiz = science_quiz

    def set_english_quiz(self,eng_score):
        self.eng_quiz=eng_score

    def set_nath_quiz(self,math_score):
        self.math_quiz=math_score

    def set_english_quiz(self,science_score):
        self.eng_quiz=science_score
    def get_name(self):
        return self.name
    def get_student_id(self):
        return self.student_id
    def get_eng_quiz(self):
        return self.eng_quiz
    def get_math_quiz(self):
        return self.math_quiz
    def get_science_quiz(self):
        return self.science_quiz 
    def get_total_score(self):
        return self.eng_quiz + self.math_quiz + self.science_quiz
    def get_avg_score(self):
         return self.get_total_score() / 3
    
    def __str__(self):
        return (f"Name : {self.name}, ID : {self.student_id}\n"
                f"English quiz score : {self.eng_quiz}, "
                f"Mathematics quiz score : {self.math_quiz}\n"
                f"Science quiz score : {self.science_quiz},\n"
                f"Total : {self.get_total_score()}, "
                f"Average : {self.get_avg_score():.1f}")
    

name = input("Enter the student’s name : ")
student_id = input("Enter the student’s ID : ")
eng_quiz = int(input("Enter the student’s English quiz score : "))
math_quiz = int(input("Enter the student’s mathematics quiz score : "))
science_quiz = int(input("Enter the student’s science quiz score : "))
student = Student(name, student_id, eng_quiz, math_quiz, science_quiz)
print(student)
```
