```python
import os

# Set the folder containing your files
folder_path = r"D:\BAS\6_sem\UAV\slides"  # Change this to your actual folder path

# Loop through all files in the folder
for filename in os.listdir(folder_path):
    if filename.lower().startswith("copy of "):  # Case insensitive check
        old_file = os.path.join(folder_path, filename)
        new_file = os.path.join(folder_path, filename[8:])  # Remove "Copy of "

        os.rename(old_file, new_file)
        print(f'Renamed: "{filename}" → "{filename[8:]}"')

print("Renaming completed!")
```

```python

import os

# Set the base directory
base_dir = r"D:\BAS\6_sem\UAV"  # Change this to your actual folder path

# Walk through all subdirectories
for root, _, files in os.walk(base_dir):
    for filename in files:
        if filename.lower().startswith("copy of "):  # Case insensitive check
            old_file = os.path.join(root, filename)
            new_file = os.path.join(root, filename[8:])  # Remove "Copy of "

            os.rename(old_file, new_file)
            print(f'Renamed: "{old_file}" → "{new_file}"')

print("Renaming completed!")
```
