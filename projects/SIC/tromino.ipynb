```python
#memorization 
memo = {0:0,1:1}
def fib2(n):
    if n in memo:
        return memo
    else:
        memo[n]= fib2(n-1)+fib2(n-2)
        return
```

```python
def pascal(n,k):
    if k ==0 or k==n:
        return 1
    else:
        return pascal(n-1,k-1)+pascal(n-1,k)
```

```python

memo = {}
def pascal2(n, k):
    if (n, k) not in memo:
        if k == 0 or k == n:
            res = 1
        else:
            res = pascal2(n-1, k-1) + pascal2(n-1, k)
        memo[(n,k)] = res
    return memo[(n,k)]
```
